```python
import seaborn as sns
import pandas as pd

taxis = sns.load_dataset("taxis")
taxis.shape
```

```python
taxis.dtypes
```

```python
taxis["pickup"] = pd.to_datetime(taxis["pickup"])
taxis["dropoff"] = pd.to_datetime(taxis["dropoff"])
taxis.dtypes
```

```python
print(taxis["pickup"].min())
print(taxis["pickup"].max())
taxis
```

```python
taxis["jour"] = taxis["pickup"].dt.date

par_jour = taxis.groupby("jour").agg(
    nb_courses=("total", "size"),
    chiffre_affaires=("total", "sum"),
).round(2)
par_jour
```

```python
courses_par_heure = taxis["pickup"].dt.hour.value_counts().sort_index()
courses_par_heure
```

```python
heure_la_plus_chargee = courses_par_heure.idxmax()
print(f"Heure la plus chargée : {heure_la_plus_chargee}h ({courses_par_heure.max()} courses)")
```

```python
noms_jours = ["Lundi", "Mardi", "Mercredi", "Jeudi", "Vendredi", "Samedi", "Dimanche"]
taxis["jour_semaine"] = taxis["pickup"].dt.dayofweek.map(lambda n: noms_jours[n])

pourboire_par_jour = taxis.groupby("jour_semaine")["tip"].mean().round(2).sort_values(ascending=False)
pourboire_par_jour
```
