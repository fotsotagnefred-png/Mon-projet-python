```python
import pandas as pd
titanic = pd.read_csv("../data/titanic_propre.csv")
titanic
```

```python
survie_par_classe_et_sexe = (titanic.groupby(["pclass", "sex"])["survived"].mean() * 100).round(1)
survie_par_classe_et_sexe
```

**Réponse 1 :** Oui, dans les trois classes. En 1re classe, 96.8 % des femmes survivent contre 36.9 % des hommes ; en 2e classe 92.1 % contre 15.7 % ; en 3e classe 50.0 % contre 13.5 %. L'écart est énorme partout.

```python
titanic["tranche_age"] = pd.cut(titanic["age"], bins=[0, 12, 17, 59, 100], labels=["Enfant 0-12", "Ado 13-17", "Adulte 18-59", "Senior 60+"])

survie_par_tranche = (titanic.groupby("tranche_age", observed=True)["survived"].mean() * 100).round(1)
survie_par_tranche
```

**Réponse 2 :** Oui. Les enfants de 0 à 12 ans survivent à 58.0 %, les ados à 47.7 %, les adultes à 36.4 % et les plus de 60 ans à 26.9 %. Plus on est jeune, plus la survie est élevée, ce qui suggère que les enfants ont été prioritaires.

```python
prix_moyen_par_port = titanic.groupby("embark_town", observed=True)["fare"].mean().round(2)
prix_moyen_par_port
```

```python
classes_par_port = (pd.crosstab(titanic["embark_town"], titanic["pclass"], normalize="index") * 100).round(1)
classes_par_port
```

**Réponse 3 :** Cherbourg a le prix moyen le plus élevé (59.95), devant Southampton (27.24) et Queenstown (13.28). La répartition des classes l'explique en grande partie : 50.6 % des passagers de Cherbourg voyagent en 1re classe, contre 20.0 % à Southampton et seulement 2.6 % à Queenstown, où 93.5 % sont en 3e classe.

```python
survie_selon_solitude = (titanic.groupby("alone")["survived"].mean() * 100).round(1)
survie_selon_solitude
```

**Réponse 4 :** Voyager seul diminuait les chances de survie : 30.4 % pour les passagers seuls contre 50.6 % pour ceux accompagnés. Attention, cela ne prouve pas que la solitude soit la cause : les passagers seuls étaient surtout des hommes adultes, un groupe qui a moins survécu.
