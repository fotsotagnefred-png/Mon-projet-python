```python
import seaborn as sns
titanic = sns.load_dataset("titanic")

nb_moins_18_ans = (titanic["age"] < 18).sum()
print(f"Moins de 18 ans : {nb_moins_18_ans}")
```

```python
billets_plus_chers = titanic.nlargest(5, "fare")[["fare", "class", "embark_town"]]
billet_le_plus_cher = billets_plus_chers["fare"].max()

print(billets_plus_chers)
print(f"Billet le plus cher : {billet_le_plus_cher:.2f}")
```

```python
premiere_classe = titanic[titanic["class"] == "First"]
part_femmes_1re = (premiere_classe["sex"] == "female").mean() * 100

print(f"Femmes en 1re classe : {part_femmes_1re:.1f} %")
```

```python
femmes_1re = titanic[(titanic["class"] == "First") & (titanic["sex"] == "female")]
survie_femmes_1re = femmes_1re["survived"].mean() * 100

hommes_3e = titanic[(titanic["class"] == "Third") & (titanic["sex"] == "male")]
survie_hommes_3e = hommes_3e["survived"].mean() * 100

print(f"Survie femmes 1re : {survie_femmes_1re:.1f} %")
print(f"Survie hommes 3e : {survie_hommes_3e:.1f} %")
```
