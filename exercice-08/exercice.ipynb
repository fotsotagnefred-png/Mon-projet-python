```python
import seaborn as sns
titanic = sns.load_dataset("titanic")
titanic.isna().sum()
```

```python
titanic
```

```python
nb_age_manquants = titanic["age"].isna().sum()

mediane_age = titanic.groupby(["class", "sex"], observed=True)["age"].transform("median")
titanic["age"] = titanic["age"].fillna(mediane_age)

titanic.isna().sum()
```

```python
nb_port_manquants = titanic["embark_town"].isna().sum()
port_le_plus_frequent = titanic["embark_town"].mode()[0]
titanic["embark_town"] = titanic["embark_town"].fillna(port_le_plus_frequent)

titanic = titanic.drop(columns="deck")

titanic.isna().sum()
```
