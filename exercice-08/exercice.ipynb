```python
import seaborn as sns
titanic = sns.load_dataset("titanic")
titanic.isna().sum()
```

```python
titanic
```

```python
nb_age_manquants = titanic["age"].isna().sum()

mediane_age = titanic.groupby(["class", "sex"], observed=True)["age"].transform("median")
titanic["age"] = titanic["age"].fillna(mediane_age)

titanic.isna().sum()
```

```python
nb_port_manquants = titanic["embark_town"].isna().sum()
port_le_plus_frequent = titanic["embark_town"].mode()[0]
titanic["embark_town"] = titanic["embark_town"].fillna(port_le_plus_frequent)

titanic = titanic.drop(columns="deck")

titanic.isna().sum()
```

```python
nb_doublons = titanic.duplicated().sum()
nb_doublons
```

```python
titanic[titanic.duplicated(keep=False)].sort_values(list(titanic.columns))
```

```python
titanic = titanic.drop(columns=["embarked", "class", "alive", "who", "adult_male"])
titanic["sex"] = titanic["sex"].astype("category")
titanic["embark_town"] = titanic["embark_town"].astype("category")

titanic.dtypes
```

```python
titanic
```

```python
titanic["fare"].describe()
```

```python
q1 = titanic["fare"].quantile(0.25)
q3 = titanic["fare"].quantile(0.75)
borne_haute = q3 + 1.5 * (q3 - q1)

billets_hors_bornes = titanic[titanic["fare"] > borne_haute].sort_values("fare", ascending=False)
nb_hors_bornes = len(billets_hors_bornes)
print(f"Borne haute : {borne_haute:.2f} / Billets hors bornes : {nb_hors_bornes}")
billets_hors_bornes
```
